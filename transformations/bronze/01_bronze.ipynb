# Task 2 - Bronze Layer
# Läser in rådata och sparar som Delta-tabell i bronze-lagret

```python
RAW_DATA_PATH = "/Volumes/marathos/default/raw/TWO_CENTURIES_OF_UM_RACES.csv"
BRONZE_TABLE  = "marathos.bronze.ultra_marathon_raw"

df_raw = (spark.read
          .option("header", "true")
          .option("inferSchema", "true")
          .csv(RAW_DATA_PATH))

print(f"Antal rader inlästa: {df_raw.count():,}")
display(df_raw.limit(5))
```

```python
from pyspark.sql.functions import col

df_renamed = df_raw.toDF(
    "year_of_event",
    "event_dates",
    "event_name",
    "event_distance_length",
    "event_number_of_finishers",
    "athlete_performance",
    "athlete_club",
    "athlete_country",
    "athlete_year_of_birth",
    "athlete_gender",
    "athlete_age_category",
    "athlete_average_speed",
    "athlete_id"
)

(df_renamed.write
           .format("delta")
           .mode("overwrite")
           .saveAsTable(BRONZE_TABLE))

print(f"Tabell sparad: {BRONZE_TABLE}")
```

```python
df_verify = spark.table(BRONZE_TABLE)
print(f"Rader i bronze-tabell: {df_verify.count():,}")
display(df_verify.limit(5))
```
