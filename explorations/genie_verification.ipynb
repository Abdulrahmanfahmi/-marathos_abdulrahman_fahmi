```python
%sql
SELECT e.event_name, COUNT(*) AS antal
FROM marathos.gold.fct_results f 
JOIN marathos.gold.dim_event e ON f.event_id = e.event_id 
GROUP BY 1 
ORDER BY 2 DESC 
LIMIT 1
```
